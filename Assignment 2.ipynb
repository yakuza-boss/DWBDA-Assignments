## APPLYING MAP REDUCE FRAMEWORK TO CALCULATE THE NUMBER OF STUDENTS IN EACH CATEGORIES (S, A, B, C, D, E, F)
## USING THE PREVIOUS SEMESTER FINAL EXAM RESULTS AS THE SAMPLE DATA
## AND ALSO TO CALCULATE FOR ALL THE SUBJECTS

```python
import re
from collections import defaultdict

# INPUT FILES

files = [
    ("PROJECT BASED LEARNING II", r"CD202A5.txt"),
    ("OBJECT ORIENTED PROGRAMMING USING JAVA LAB", r"CD203A4.txt"),
    ("DESIGN AND ANALYSIS OF ALGORITHMS", r"CD204A1.txt"),
    ("DATABASE MANAGEMENT SYSTEMS LAB", r"CD204A4.txt"),
    ("DATABASE MANAGEMENT SYSTEMS", r"CD205A1.txt"),
    ("PROGRAMMING IN JAVA", r"CD207A3.txt"),
    ("UNIVERSAL HUMAN VALUES II", r"GN201B1.txt"),
    ("PROBABILITY, STATISTICS AND STOCHASTIC PROCESSES", r"MA206B1.txt")
]

grades = ["S", "A", "B", "C", "D", "E", "F"]

# MAPPING

mapped_data = []

for subject, file in files:

    with open(file, "r", encoding="utf-8") as f:

        for line in f:

            line = line.strip()

            # Extract grade from the end of each student record
            match = re.search(r'\b(S|A|B|C|D|E|F)\s*$', line)

            if match:

                grade = match.group(1)

                # Mapper output
                mapped_data.append(
                    (subject, grade, 1)
                )

# SHUFFLE AND SORT

grouped_data = defaultdict(list)

for subject, grade, value in mapped_data:

    grouped_data[(subject, grade)].append(value)

# REDUCE

result = defaultdict(dict)

for (subject, grade), values in grouped_data.items():

    result[subject][grade] = sum(values)

# DISPLAY RESULT

print("\n")
print("=" * 108)
print("                                                        MAPREDUCE STUDENT GRADE COUNT")
print("=" * 108)

print(
    f"{'Subject':<50}"
    f"{'S':>7}"
    f"{'A':>7}"
    f"{'B':>7}"
    f"{'C':>7}"
    f"{'D':>7}"
    f"{'E':>7}"
    f"{'F':>7}"
    f"{'Total':>9}"
)

print("-" * 108)


for subject, file in files:

    counts = []

    for grade in grades:

        counts.append(
            result[subject].get(grade, 0)
        )

    total = sum(counts)

    print(
        f"{subject:<50}"
        f"{counts[0]:>7}"
        f"{counts[1]:>7}"
        f"{counts[2]:>7}"
        f"{counts[3]:>7}"
        f"{counts[4]:>7}"
        f"{counts[5]:>7}"
        f"{counts[6]:>7}"
        f"{total:>9}"
    )

print("=" * 108)
```
